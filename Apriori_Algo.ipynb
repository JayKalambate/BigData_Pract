```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

#Sample Data

dataset = [
    ['Milk','Bread','Butter'],
    ['Bread','Butter'],
    ['Milk','Bread'],
    ['Milk','Butter'],
    ['Bread','Butter']
]

#Convert to one-hot Encoding
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

# Step 1 : Apply Apriori

frequent_itemsets = apriori(df, min_support = 0.6, use_colnames = True)

#Step 2: Generate Rules
rules = association_rules(frequent_itemsets, metric = "confidence", min_threshold = 0.7)
print("Frequent Itemsets: ")
print(frequent_itemsets)

print("\n Association Rules: ")
print(rules[['antecedents','consequents','support','confidence']])
```

```python
import hashlib

#Sample transaction data

transactions = [
    ['Milk','Bread','Butter'],
    ['Bread','Butter'],
    ['Milk','Bread'],
    ['Milk','Butter'],
    ['Bread','Butter']
]
#set to store unique transaction hashes
unique_hashes = set()

print("Unique Transactions: ")
for txn in transactions:
    txn_sorted = sorted(txn)
    txn_string = ','.join(txn_sorted)

    #Create hash value using MD5
    txn_hash = hashlib.md5(txn_string.encode()).hexdigest()

    if txn_hash not in unique_hashes:
        unique_hashes.add(txn_hash)
        print(txn_sorted)
```
