```python
# Naive Bayes theorem 

from sklearn.naive_bayes import GaussianNB

#Features:
#Weight (Grams), Texture(1 = Smooth, 0 = Rough)

x = [
    [150,1], # Apple
    [170, 1], # Apple
    [140, 1], # Apple
    [130, 0], #Orange
    [120, 0], #Orange
    [135, 0] # Orange 
]

# Labels
# 0 = Apple
# 1 = Orange

y = [0,0,0,1,1,1]

# Create Model
model = GaussianNB

# Train Model
model.fit(x,y)

# Predict new fruit 
new_fruit = [[130,0]]

prediction = model.predict(new_fruit)

if prediction[0] == 0:
    print("The Fruit is Apple : ")
else:
    print("The Fruit is orange")
    
```
